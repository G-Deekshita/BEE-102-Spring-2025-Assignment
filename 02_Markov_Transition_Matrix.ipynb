# **Building Markov transition matrix**

You have learned important differences between simple dinucleotide frequency model and Markov models. Your goal is to build one from the input sequence. In the Jupyter notebook, you should print the transition matrix.

# **Approach**

**1. Define the states**

The four valid nucleotides in DNA are: 'A', 'C', 'G', 'T'.

These are the possible states in our Markov model.

**2. Count transitions (dinucleotides)**

For a first-order Markov model, we assume the probability of each nucleotide depends only on the one immediately before it.

So, we iterate through the sequence, looking at each pair of consecutive nucleotides.

For every such pair (current, next), we increment a counter in a nested dictionary transition_counts[current][next].

**3. Normalize to get transition probabilities**

Once all transitions are counted, we convert counts to probabilities:

For each row (starting base), sum all transitions.

Divide each count in the row by this sum to get the transition probability to the next base.

This results in a row-stochastic matrix, where each row sums to 1.

**4. Store and display the matrix**

The final probability matrix is stored in a pandas.DataFrame, where:

Rows represent the current base.

Columns represent the next base.

Each cell contains the probability of transitioning from one nucleotide to another.

```python
import pandas as pd

# Given DNA sequence for analysis
dna_sequence = "AGTTCAGTTATTAATCTGGCTTATATACTGACTACTATATCTCTTATATATACT"

# Valid nucleotide states
nucleotides = ['A', 'C', 'G', 'T']
```

```python
# Step 1: Count transitions (dinucleotides)
transition_counts = {
    base: {next_base: 0 for next_base in nucleotides}
    for base in nucleotides
}

# Step 2: Count each base-to-base transition
for idx in range(len(dna_sequence) - 1):
    current = dna_sequence[idx]
    next_base = dna_sequence[idx + 1]
    if current in nucleotides and next_base in nucleotides:
        transition_counts[current][next_base] += 1
    else:
        print(f"Skipping invalid base: {current} or {next_base}")
```

```python
# Step 3: Normalize counts to probabilities
markov_probabilities = {
    base: {} for base in nucleotides
}

for base in nucleotides:
    total = sum(transition_counts[base].values())
    for next_base in nucleotides:
        prob = transition_counts[base][next_base] / total if total > 0 else 0.0
        markov_probabilities[base][next_base] = prob
```

```python
# Step 4: Create and print the transition matrix
transition_df = pd.DataFrame(markov_probabilities)

print("Markov Transition Matrix (1st-order):")
print("Rows: Current base, Columns: Next base")
print(transition_df.round(5))
```
